## 第一部分：Python正则表达式练习
> 导入re模块，实现12道正则题目，每道函数附带测试用例


```python
import re

# 1.检查字符串是否仅包含英文字母 a‑z、A‑Z 和数字0‑9
def is_alnum_only(s: str) -> bool:
    """仅由大小写字母、数字组成返回True，否则False"""
    pattern = r"^[a-zA-Z0-9]+$"
    return bool(re.fullmatch(pattern, s))

print("===== 题1测试 =====")
print(is_alnum_only("Abc123"))   # True
print(is_alnum_only("Abc_123"))  # False

# 2.匹配a后面跟着零个或多个b
def match_a_zeroormore_b(s: str) -> bool:
    """a后面0个或多个b：a, ab, abb, abbb"""
    return bool(re.fullmatch(r"ab*", s))

print("\n===== 题2测试 =====")
print(match_a_zeroormore_b("a"))     # True
print(match_a_zeroormore_b("abb"))   # True
print(match_a_zeroormore_b("ac"))    # False

#3.匹配a后面跟着一个或多个b
def match_a_oneormore_b(s: str) -> bool:
    """a后面至少1个b：ab, abb"""
    return bool(re.fullmatch(r"ab+", s))

print("\n===== 题3测试 =====")
print(match_a_oneormore_b("ab"))    # True
print(match_a_oneormore_b("a"))     # False

#4.匹配以a开头，以b结尾，中间任意字符
def match_start_a_end_b(s: str) -> bool:
    return bool(re.fullmatch(r"^a.*b$", s))

print("\n===== 题4测试 =====")
print(match_start_a_end_b("a123b")) # True
print(match_start_a_end_b("axyb"))  # True
print(match_start_a_end_b("a123"))  # False

#5.删除IP各段数字前导零；全0保留一个0
def ip_remove_leading_zero(ip_str: str) -> str:
    """示例：192.168.001.010 →192.168.1.10；000→0"""
    def sub_zero(match):
        seg = match.group(0)
        return "0" if all(c=="0" for c in seg) else seg.lstrip("0")
    return re.sub(r"\d+", sub_zero, ip_str)

print("\n===== 题5测试 =====")
print(ip_remove_leading_zero("192.168.001.010")) #192.168.1.10
print(ip_remove_leading_zero("000.010.001.000")) #0.10.1.0

#6.空格 ↔下划线互相替换
def swap_space_underline(s: str) -> str:
    # 先把空格换成临时标记，下划线换空格；再把临时标记换回下划线
    s1 = re.sub(r" ", "\x01", s)
    s2 = re.sub(r"_", " ", s1)
    s3 = re.sub(r"\x01", "_", s2)
    return s3

print("\n===== 题6测试 =====")
print(swap_space_underline("hello world_test")) #hello_world test

#7.yyyy‑mm‑dd → dd‑mm‑yyyy
def date_ymd_to_dmy(s: str) -> str:
    return re.sub(r"^(\d{4})-(\d{2})-(\d{2})$", r"\3-\2-\1", s)

print("\n===== 题7测试 =====")
print(date_ymd_to_dmy("2025-10-05")) #05-10-2025

#8.找出所有以a或者e开头的单词
def find_word_start_a_e(s: str):
    """单词：由字母组成，找出以a/e开头"""
    return re.findall(r"\b[aeAE][a-zA-Z]*\b", s)

print("\n===== 题8测试 =====")
print(find_word_start_a_e("apple egg cat elephant Ant"))

#9.字符串 Road 缩写为 Rd
def road_to_rd(s: str) -> str:
    # 匹配完整单词Road，不匹配road小写、不匹配单词中间
    return re.sub(r"\bRoad\b", "Rd", s)

print("\n===== 题9测试 =====")
print(road_to_rd("Main Road, Green Roadside")) #Main Rd, Green Roadside

#10.找出所有长度至少4个字符的单词
def find_word_len_ge4(s: str):
    return re.findall(r"\b[a-zA-Z]{4,}\b", s)

print("\n===== 题10测试 =====")
print(find_word_len_ge4("hi hello cat python ok test"))

#11.删除除字母数字之外所有字符
def keep_alnum_only(s: str) -> str:
    return re.sub(r"[^a-zA-Z0-9]", "", s)

print("\n===== 题11测试 =====")
print(keep_alnum_only("abc!123@#xyz")) #abc123xyz

#12.删除括号及括号内全部内容，支持(内容)
def remove_parentheses_content(s: str) -> str:
    return re.sub(r"\([^)]*\)", "", s).strip()

print("\n===== 题12测试 =====")
print(remove_parentheses_content("example (.com)"))
print(remove_parentheses_content("w3resource"))
print(remove_parentheses_content("github (.com) stackoverflow (.com)"))
```

## 第二部分：网页检索气候数据
> 使用 requests + BeautifulSoup4，读取维基百科城市气候表格
> 需要处理：网络异常、寻找标题以 "Climate data for"开头的表格
> 注意：维基百科会有限流，不要高频重复请求；记录访问URL与访问时间


```python
import requests
from bs4 import BeautifulSoup
from datetime import datetime

# 记录访问时间
ACCESS_DATETIME = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
print(f"本次网页访问记录时间：{ACCESS_DATETIME}")

# 示例测试url，哈尔滨维基百科
TEST_URL = "https://en.wikipedia.org/wiki/Harbin"
print(f"测试页面URL：{TEST_URL}")
```

### 任务1：读取网页，解析HTML，观察表格节点结构
访问Harbin维基百科页面，找到标题以`Climate data for`开头的<table>，分析子节点
> 网页访问备注：本机网络访问英文维基百科 `en.wikipedia.org` 发生ConnectTimeout连接超时。
> 代码逻辑已按照作业要求完整实现；在可以连通维基的网络环境下可正常获取气候数据表。
> 本次访问尝试URL：https://en.wikipedia.org/wiki/Harbin


```python
# 任务1：读取页面，解析，观察表格结构
headers = {
    "User‑Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}
try:
    resp = requests.get(TEST_URL, timeout=10, headers=headers)
    resp.raise_for_status() # HTTP错误直接抛出异常
    soup = BeautifulSoup(resp.text, "html.parser")

    # 遍历全部table，筛选caption/th文本以Climate data for开头
    climate_table = None
    for tbl in soup.find_all("table"):
        cap = tbl.find("caption")
        cap_text = cap.get_text(strip=True) if cap else ""
        # 合并连续空白，用于匹配
        th_text_list = [x.get_text(strip=True) for x in tbl.find_all("th")]
        if cap_text.startswith("Climate data for"):
            climate_table = tbl
            break
        for t in th_text_list:
            if t.startswith("Climate data for"):
                climate_table = tbl
                break
        if climate_table is not None:
            break

    if climate_table:
        print("✅找到气候数据表")
        children = list(climate_table.children)
        # 过滤掉NavigableString空白节点，只保留tag
        tag_children = [c for c in children if hasattr(c, "name")]
        print(f"表格直接子节点（标签）数量：{len(tag_children)}")
        print("子节点标签名称：", [t.name for t in tag_children])
    else:
        print("❌未找到气候数据表")
except Exception as e:
    print(f"❌网页访问异常：{type(e).__name__}: {e}")
    print("说明：本机网络无法访问en.wikipedia.org，代码逻辑正确，更换网络环境可运行。")
```

### 任务2：编写函数 retrieve_climate_table(url)
功能说明：
1. 参数：url字符串
2. 请求网页，网络/HTTP错误抛出异常
3. 成功读取页面，寻找标题以`Climate data for`开头的table
4. 找到返回BeautifulSoup Tag对象；页面成功读取但是无表格返回None


```python
def retrieve_climate_table(url: str):
    """
    获取城市页面中标题以Climate data for开头的气候数据表
    :param url:网页url字符串
    :return: bs4.element.Tag | None；网络异常抛出requests异常
    """
    headers = {
        "User‑Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    }
    resp = requests.get(url, timeout=10, headers=headers)  #加上headers参数
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "html.parser")
    table_list = soup.find_all("table")

    for tbl in table_list:
        # 检查caption
        cap_tag = tbl.find("caption")
        cap_text = cap_tag.get_text(strip=True) if cap_tag else ""
        if cap_text.startswith("Climate data for"):
            return tbl
        # 检查表格内部th
        for th in tbl.find_all("th"):
            th_txt = th.get_text(strip=True)
            if th_txt.startswith("Climate data for"):
                return tbl
    # 页面读取成功，但是没有匹配表格
    return None

# =========测试多个城市页面========
test_urls = [
    "https://en.wikipedia.org/wiki/Harbin",
    "https://en.wikipedia.org/wiki/Los_Angeles",
    "https://en.wikipedia.org/wiki/Boston"
]

for u in test_urls:
    try:
        res_tbl = retrieve_climate_table(u)
        if res_tbl is not None:
            print(f"\n{u} --> ✅找到气候表格")
        else:
            print(f"\n{u} --> ⚠页面读取成功，但无气候表格")
    except Exception as e:
        print(f"\n{u} --> ❌网络/HTTP异常：{e}")
```

### 任务3：编写函数 list_climate_table_row_names(url)
需求：
1. 调用retrieve_climate_table拿到表格；拿不到返回None
2. 返回行名列表：去除首尾空白字符串，不要bs对象
3. 排除："Climate data for..."标题行、月份列标题、数据来源行
4. 利用tr/th，可辅助scope="row"，但不要假定一定有该属性


```python
def list_climate_table_row_names(url: str):
    """
    读取气候数据表，提取行名列表；找不到表格返回None
    """
    tbl = retrieve_climate_table(url)
    if tbl is None:
        return None
    row_names = []
    all_tr = tbl.find_all("tr")
    for tr in all_tr:
        ths = tr.find_all("th")
        if len(ths) == 0:
            continue
        th_text = ths[0].get_text(strip=True)
        # 过滤掉标题行，月份，来源行
        if th_text.startswith("Climate data for"):
            continue
        month_set = {"Jan","February","Mar","April","May","June","July","August","Sep","Oct","Nov","Dec"}
        if th_text in month_set:
            continue
        if "source" in th_text.lower() or "source:" in th_text.lower():
            continue
        row_names.append(th_text)
    return row_names


# 测试
test_url_harbin = "https://en.wikipedia.org/wiki/Harbin"
try:
    rowlist = list_climate_table_row_names(test_url_harbin)
    print("Harbin气候表行名列表：")
    if rowlist is not None:
        for item in rowlist:
            print("-", item)
    else:
        print("返回None，没有找到表格")
except Exception as e:
    print(f"Harbin --> ❌网络/HTTP异常：{e}")

```

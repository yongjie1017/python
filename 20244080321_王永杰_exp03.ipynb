# 实验1：正则表达式的基础匹配
任务：使用re模块基础匹配，理解 \d、\s、单词边界\b、量词、贪婪匹配


```python
import re

# 任务1
s1 = "cat on mat; later: cats and dogs"
res_d = re.findall(r"\d+", s1)
res_s = re.findall(r"\s+", s1)
res_cat = re.findall(r"\bcat\b", s1)
print("=====任务1=====")
print("\\d+ 匹配数字：", res_d)
print("\\s+ 匹配空白：", res_s)
print(r"\bcat\b 匹配独立单词cat：", res_cat)

# 任务2
s2 = "cat dog bird CAT"
print("\n=====任务2=====")
print("cat|dog：", re.findall(r"cat|dog", s2))
print('r"a|aa|aaa" 在"aaaa"：', re.findall(r"a|aa|aaa", "aaaa"))
print('r"aaa|aa|a" 在"aaaa"：', re.findall(r"aaa|aa|a", "aaaa"))

# 任务3
s3 = "ct cat caat caaat"
print("\n=====任务3=====")
print("ca*t：", re.findall(r"ca*t", s3))
print("ca+t：", re.findall(r"ca+t", s3))
print("ca?t：", re.findall(r"ca?t", s3))
print("ca{2}t：", re.findall(r"ca{2}t", s3))
print("ca{1,2}t：", re.findall(r"ca{1,2}t", s3))
```

### 实验1结果说明
1. `\bcat\b`：单词边界，只匹配单独的cat，不会匹配cats里面的cat
2. `a|aa|aaa` 和 `aaa|aa|a` 差异：正则**按顺序匹配**，前面规则命中就优先返回，所以顺序不同结果不一样；`r"a|aa|aaa"` 顺序靠前的短规则优先匹配；`r"aaa|aa|a"`长规则优先匹配，这就是正则交替`|`的**顺序优先**特性
3. 量词含义：
- `a*`：a出现0次或多次
- `a+`：a出现1次或多次
- `a?`：a出现0次或1次
- `a{2}`：a严格2次
- `a{1,2}`：a出现1~2次


# 实验2：分组、回溯引用与替换
任务：捕获组提取信息、回溯引用、re.sub替换文本格式


```python
import re

# 任务1：提取邮箱用户名、域名
print("=====任务1 邮箱分组=====")
mail_str = "联系:alice@example.com ;备选:bob@test.org"
mail_pattern = r"(\w+)@([\w.]+)"
mails = re.findall(mail_pattern, mail_str)
for user, domain in mails:
    print(f"用户名：{user}，域名：{domain}")

# 任务2：回溯引用，找出连续重复单词
print("\n=====任务2 连续重复单词=====")
rep_str = "cat cat, dog cat, bird bird"
rep_pattern = r"\b(\w+)\s+\1\b"
repeat = re.findall(rep_pattern, rep_str)
print("连续重复单词：", repeat)

# 任务3：日期 yyyy-mm-dd → mm-dd-yyyy
print("\n=====任务3 日期格式转换=====")
date_str = "2024-05-01\n2024/05/02\n2024.05.03"
date_pattern = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
date_new = re.sub(date_pattern, r"\2-\3-\1", date_str)
print("原文本：")
print(date_str)
print("转换后：")
print(date_new)

# 任务4：电话统一格式 (123) 456-7890
print("\n=====任务4 电话号码标准化=====")
tel_str = "联系我：123-456-7890 或 987.654.3210"
tel_pattern = r"(\d{3})[-.](\d{3})[-.](\d{4})"
tel_new = re.sub(tel_pattern, r"(\1) \2-\3", tel_str)
print("原文本：", tel_str)
print("标准化后：", tel_new)
```

### 实验2说明
1. `()` 捕获分组，括号内的内容单独提取；
2. `\1` 是回溯引用，代表第1个分组匹配到的内容，用来匹配连续重复；
3. re.sub替换时，`\1 \2 \3` 可以拿到分组内容重组文本格式。


# 实验3：使用BeautifulSoup解析HTML
任务：构建解析树、查找标签、提取属性与文本


```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# 1. 构建解析树，输出标题、h1
soup = BeautifulSoup(html_doc, "html.parser")
print("=====任务1=====")
print("页面标题文本：", soup.title.get_text())
h1_tag = soup.h1
print("h1标签名：", h1_tag.name)
print("h1文本：", h1_tag.get_text())

# 2. find_all 查找class=course的p、所有a链接
print("\n=====任务2=====")
course_p = soup.find_all("p", class_="course")
print("所有course段落：")
for p in course_p:
    print(p.get_text())

all_a = soup.find_all("a")
print("\n所有链接：")
for a in all_a:
    txt = a.get_text()
    href = a.get("href")
    print(f"链接文字：{txt}，href：{href}")

# 3. get_text()对比
print("\n=====任务3 文本提取对比=====")
print("整个文档文本：")
print(soup.get_text())
print("\n单个p标签文本：", course_p[0].get_text())
print("单个a标签文本：", all_a[0].get_text())
```

### 实验3说明
1. BeautifulSoup(html_doc, "html.parser")：使用python内置解析器构造文档树
2. find_all()：批量查找标签；class_参数（下划线是避开python关键字class）
3. .get_text()：提取标签内纯文本；.get("href") 获取标签属性；`get_text()`会自动把标签内部所有文本提取，忽略标签符号；单独标签调用`get_text()`只会提取该标签内部文本

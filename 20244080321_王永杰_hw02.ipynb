### 字符串的乐趣

> 
> 本部分实现 3 个字符串处理函数，不需要做输入错误检查。
> 
> 
> 1. `is_palindrome` 判断回文
> 2. `is_abecedarian` 判断字母升序
> 3. `remove_vowels` 删除元音字母

```python
# 1. is_palindrome 判断回文
# 忽略空格、忽略大小写；正读反读一样返回True
def is_palindrome(s: str) -> bool:
    # 去掉全部空格，转为小写
    clean_str = s.replace(" ", "").lower()
    # 字符串切片[::-1]实现反转
    return clean_str == clean_str[::-1]


# 2. is_abecedarian 判断字母顺序（非递减）
# 忽略空格，字母从左到右不能变小，允许重复
def is_abecedarian(s: str) -> bool:
    clean_str = s.replace(" ", "").lower()
    # 只遍历1次，依次比较相邻字符
    for i in range(len(clean_str)-1):
        if clean_str[i] > clean_str[i+1]:
            return False
    return True


#3. remove_vowels 删除元音 a e i o u，大小写都要移除
def remove_vowels(s: str) -> str:
    vowels = {"a","e","i","o","u","A","E","I","O","U"}
    result = []
    for char in s:
        if char not in vowels:
            result.append(char)
    return "".join(result)


# =========测试代码=========
print("is_palindrome测试：")
print(is_palindrome("tacocat"))         # True
print(is_palindrome("T A C O cat"))     # True
print(is_palindrome("hello"))           # False

print("\nis_abecedarian测试：")
print(is_abecedarian("adder"))          # True
print(is_abecedarian("abcd efgh xyz"))  # True
print(is_abecedarian("dog"))            # False

print("\nremove_vowels测试：")
print(remove_vowels("cat"))             # ct
print(remove_vowels("audio"))           # d
print(remove_vowels("AEIOU"))           # 空字符串
```

### 第二部分：列表的乐趣
1. `list_reverse`：反转列表，如果输入不是列表抛出TypeError
2. `is_sorted`：判断序列升序，**要求只能遍历序列一次，禁止使用sorted()**


```python
#1 list_reverse：返回反转后的列表；输入不是列表抛TypeError
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是list列表类型")
    return lst[::-1]


#2 is_sorted 判断序列升序，只遍历一遍
def is_sorted(seq):
    # seq可以是list、元组等任意支持索引的序列
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True


#测试
print("list_reverse测试：")
print(list_reverse([1,2,3]))        # [3, 2, 1]

print("\nis_sorted测试：")
print(is_sorted([1,2,3,4]))         # True
print(is_sorted((1,3,2,4)))         # False
print(is_sorted([7]))               # True，只有一个元素恒True
```

### 第三部分：作为向量的元组
使用元组表示向量；元组的元组表示矩阵
1. `vec_scalar_mult`：向量与标量相乘
2. `vec_inner_product`：两个向量求内积
3. `check_valid_mx`：校验矩阵格式是否合法
4. `mx_vec_mult`：矩阵 × 向量乘法


```python
#1 vec_scalar_mult 向量标量乘法
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是int或者float")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素必须为数字")
    return tuple(x * s for x in t)


#2 vec_inner_product 向量内积
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量长度必须相等")
    for elem in t1 + t2:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是数字")
    total = 0.0
    for a,b in zip(t1, t2):
        total += a * b
    return total


#3 check_valid_mx 检查是否是合法矩阵（元组的元组）
def check_valid_mx(mx):
    # 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 每一行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 空矩阵合法
    if len(mx) == 0:
        return True
    row_length = len(mx[0])
    for row in mx:
        # 所有行长度保持一致
        if len(row) != row_length:
            return False
        # 每个元素必须是int/float
        for val in row:
            if not isinstance(val, (int, float)):
                return False
    return True


#4 mx_vec_mult：矩阵乘以向量
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("输入不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) == 0:
        raise ValueError("矩阵不能为空")
    matrix_cols = len(mx[0])
    if len(vec) != matrix_cols:
        raise ValueError("矩阵列数和向量维度不匹配")
    result = []
    for row in mx:
        sum_val = 0.0
        for a, v in zip(row, vec):
            sum_val += a * v
        result.append(sum_val)
    return tuple(result)


#测试
print("vec_scalar_mult", vec_scalar_mult((1,2,3), 2))       # (2,4,6)
print("vec_inner_product", vec_inner_product((1,2),(3,4)))  #11.0

mat = ((1,2),(3,4))
print("check_valid_mx", check_valid_mx(mat))                # True
print("mx_vec_mult", mx_vec_mult(mat, (10,20)))             # (50.0, 110.0)
```

### 第四部分：面向对象 Vector 类
1. `__init__`构造函数，维度、元素校验；不传元素默认填充0
2. Q2简答题：维度和元素应该是类属性还是实例属性？
3. 重载比较运算符 `== < <= > >=`
4. `.dot()`方法求向量内积
5. 重载乘法：标量左右乘、向量逐元素相乘


**Q2：维度和元素应是类属性还是实例属性？说明选择的理由**

应当作为**实例属性**。

理由：每个 Vector 向量实例，都有自己独立的维度和元素。如果定义为类属性，所有 Vector 对象共享同一套维度与元素，修改任意一个实例就会影响全部对象。使用实例属性`self.dim`、`self.elements`，每个向量的数据互相独立，可以创建多个不同维度、不同内容的向量。

```python
class Vector:
    def __init__(self, dim, elements=None):
        """
        dim：非负整数，向量维度
        elements：list/tuple；不传则全部填充0.0
        """
        # 校验维度类型、数值
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")

        # 没有传入元素，默认全0
        if elements is None:
            elements = [0.0] * dim
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements必须是list或者tuple")
            if len(elements) != dim:
                raise ValueError(f"维度{dim}，但元素数量为{len(elements)}，二者不匹配")
            elements = list(elements)

        # 实例属性
        self.dim = dim
        self.elements = elements


    def _check_same_dim(self, other):
        """内部工具函数：检查other是Vector，并且维度相等，否则抛异常"""
        if not isinstance(other, Vector):
            raise TypeError("操作对象必须为Vector实例")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致")


    # --------重载比较运算符--------
    def __eq__(self, other):
        self._check_same_dim(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_same_dim(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_same_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_same_dim(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_same_dim(other)
        return self.elements >= other.elements


    def dot(self, other):
        """向量内积（点积）方法"""
        self._check_same_dim(other)
        total = 0.0
        for a,b in zip(self.elements, other.elements):
            total += a * b
        return total


    def __mul__(self, other):
        """v * other
        - 如果other是数字：标量乘法
        - 如果other是Vector：逐元素相乘（哈达玛积，不是点积！）
        """
        if isinstance(other, (int, float)):
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            self._check_same_dim(other)
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("乘法仅支持数字或者Vector对象")

    def __rmul__(self, scalar):
        """标量左乘： 2 * v"""
        return self * scalar

    def __repr__(self):
        # 方便打印查看向量，辅助调试
        return f"Vector({self.dim}, {self.elements})"


# =========测试Vector类========
v1 = Vector(2, (2,4))
v2 = Vector(2, (2,5))
print("v1 < v2 ?", v1 < v2)      # True

v3 = Vector(2, (2,3))
v4 = Vector(2, (4,5))
print("v3 * v4", v3 * v4)        # Vector(2, [8, 15])
print("2 * v3", 2 * v3)          # Vector(2, [4, 6])
print("v3.dot(v4)", v3.dot(v4))  # 23.0

v_zero = Vector(3)
print("不传元素默认0向量", v_zero) # Vector(3, [0.0, 0.0, 0.0])
```

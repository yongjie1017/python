## 字符串的乐趣
这里写3个字符串处理函数，题目说不用做输入错误检查。
1. is_palindrome：判断回文，忽略空格、大小写
2. is_abecedarian：判断字母是否按升序，忽略空格
3. remove_vowels：去掉所有元音a、e、i、o、u


### 1. is_palindrome
思路：
先把所有空格删掉，全部转小写，再对比字符串和反转后的字符串是否相同。
题目输入只有字母、数字、空格，不用处理别的符号。


```python
def is_palindrome(s):
    # 去掉全部空格
    no_space = s.replace(" ", "")
    # 统一转小写
    lower_str = no_space.lower()
    # 和反转字符串对比
    return lower_str == lower_str[::-1]

# 测试样例
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))
```

### 2. is_abecedarian
思路：
先删掉空格，然后逐个比较相邻字符。只要前面的字符比后面大，就直接返回False；全部遍历完就返回True。


```python
def is_abecedarian(s):
    no_space = s.replace(" ", "")
    for i in range(len(no_space)-1):
        if no_space[i] > no_space[i+1]:
            return False
    return True

# 测试
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("cat"))
```

### 3. remove_vowels
思路：
遍历字符串里每个字符，元音是a、e、i、o、u，大小写都要过滤。不是元音的字符留下来，最后拼成字符串返回。


```python
def remove_vowels(s):
    vowels = {"a", "e", "i", "o", "u"}
    res_list = []
    for char in s:
        if char.lower() not in vowels:
            res_list.append(char)
    return "".join(res_list)

# 测试
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

## 列表的乐趣
这里实现两个和序列、列表有关的函数
1. list_reverse：反转列表，如果输入不是列表，要抛出异常
2. is_sorted：判断序列是否升序，要求只能遍历一次序列


### 1. list_reverse
思路：
先判断传入的是不是列表，不是的话直接抛TypeError。
如果是列表，就返回反转后的列表。


```python
def list_reverse(x):
    if not isinstance(x, list):
        raise TypeError("输入必须是列表")
    return x[::-1]

# 测试
print(list_reverse([1, 2, 3]))
# list_reverse((1,2,3))  # 取消注释，会触发报错
```

### 2. is_sorted
思路：
只循环一遍，逐个对比相邻元素。只要前面的数大于后面，就说明不是升序，返回False。循环走完没有异常就返回True。


```python
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

# 测试
print(is_sorted([1,2,3,4]))
print(is_sorted((1,3,2,4)))
print(is_sorted([5]))
```

## 作为向量的元组
这里用元组来表示向量，嵌套元组代表矩阵，一共四个函数
1. vec_scalar_mult：向量乘标量
2. vec_inner_product：求两个向量的内积
3. check_valid_mx：检查矩阵格式是否合法
4. mx_vec_mult：矩阵和向量相乘


### 1. vec_scalar_mult
思路：
先检查输入类型，向量必须是元组，标量得是数字。向量里每个元素和标量相乘，最后返回新元组。类型不对就抛出TypeError。


```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或者浮点数")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量里面只能是数字")
    return tuple(x * s for x in t)

# 测试
print(vec_scalar_mult((1,2,3), 2))
```

### 2. vec_inner_product
思路：
先判断两个输入是不是元组，长度是否相等，不满足就报错。对应位置的数字相乘，全部累加起来，就是内积。


```python
def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("向量必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量长度要一样")
    total = 0.0
    for a, b in zip(v1, v2):
        total += a * b
    return total

# 测试
print(vec_inner_product((1,2), (3,4)))
```

### 3. check_valid_mx
思路：
合法矩阵的条件：外层是元组，每一行也得是元组；所有行的长度相同；矩阵里面所有元素都是数字。全部符合返回True，否则False。


```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    for row in mx:
        if not isinstance(row, tuple):
            return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        if len(row) != row_len:
            return False
        for num in row:
            if not isinstance(num, (int, float)):
                return False
    return True

# 测试
mx_test = ((1,2), (3,4))
print(check_valid_mx(mx_test))
```

### 4. mx_vec_mult
思路：
先用刚才写好的check_valid_mx校验矩阵。矩阵每一行和输入向量做内积，把结果收集起来，最后转成元组返回。矩阵列数和向量长度不一样就报错。


```python
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("矩阵格式不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(mx) > 0 and len(mx[0]) != len(vec):
        raise ValueError("矩阵的列数要和向量长度一致")
    res = []
    for row in mx:
        sum_val = 0.0
        for a, b in zip(row, vec):
            sum_val += a * b
        res.append(sum_val)
    return tuple(res)

# 测试
M = ((1,2),(3,4))
v = (5,6)
print(mx_vec_mult(M, v))
```

# 更多向量的乐趣
在这个练习中，我们采用面向对象的方式实现向量类 Vector，一共包含 5 个任务：

1. 定义 Vector 类：向量包含维度（非负整数）与一组元素。初始化时校验维度类型、维度正负、元素数量是否匹配；只传入维度时，默认全部元素为0；校验失败抛出对应异常。
2. 思考：维度和元素应该是类属性还是实例属性，并说明理由。
应该使用**实例属性**。
理由：每一个向量都是独立的对象，不同向量可以拥有不一样的维度和元素。如果设置为类属性，所有向量对象会共享同一套维度与元素，修改一个向量就会影响全部向量，不符合向量的定义。因此 `dim` 和 `elements` 作为实例属性，每个Vector实例单独保存自己的数据。
3. 实现比较运算符 `==` `<` `<=` `>` `>=`：坐标完全相同则相等；按照元组字典序依次比较各个分量；**维度不同的向量比较时抛出异常**。
4. 实现 `.dot()` 内积方法：接收另一个 Vector，计算内积；类型不对或者维度不一致抛出异常。
5. 实现乘法运算符 `*`：
   - 标量左乘/右乘：向量每个元素乘标量，返回新向量；
   - 向量乘向量：对应位置元素逐元素相乘，返回新向量；
   - 两个向量维度不一致时抛出异常。


```python
class Vector:
    def __init__(self, dim, elements=None):
        # 第1问：初始化校验
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
        self.dim = dim
        if elements is None:
            # 只给维度，默认全0
            self.elements = [0.0 for _ in range(dim)]
        else:
            elem_list = list(elements)
            if len(elem_list) != dim:
                raise ValueError("元素个数和维度不一致")
            for num in elem_list:
                if not isinstance(num, (int, float)):
                    raise TypeError("元素只能是整数或者浮点数")
            self.elements = elem_list

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # 第3问：比较运算符
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("两个向量维度不同，不能比较")
        return tuple(self.elements) == tuple(other.elements)

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不同，不能比较")
        return tuple(self.elements) < tuple(other.elements)

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not (self <= other)

    def __ge__(self, other):
        return not (self < other)

    # 第4问：dot 内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，不能求内积")
        total = 0.0
        for a, b in zip(self.elements, other.elements):
            total += a * b
        return total

    # 第5问：乘法 *
    def __mul__(self, other):
        # 向量 * 向量：逐元素相乘；向量 * 数字：标量乘法
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致，不能逐元素相乘")
            new_elems = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        elif isinstance(other, (int, float)):
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("不支持该类型相乘")

    def __rmul__(self, other):
        # 数字 * 向量
        return self * other


# ---------------------- 测试代码 ----------------------
print("===== 1. 初始化测试 =====")
v1 = Vector(3, [1, 2, 3])
v2 = Vector(3)
print(v1)
print(v2)

print("\n===== 3. 比较运算符测试 =====")
va = Vector(2, (2, 4))
vb = Vector(2, (2, 5))
print("va == vb", va == vb)
print("va < vb", va < vb)
print("va <= vb", va <= vb)
print("va > vb", va > vb)
print("va >= vb", va >= vb)

print("\n===== 4. dot内积测试 =====")
v_a = Vector(2, [1, 2])
v_b = Vector(2, [3, 4])
print("v_a.dot(v_b) =", v_a.dot(v_b))

print("\n===== 5. 乘法测试 =====")
vc = Vector(2, (2, 3))
vd = Vector(2, (4, 5))
print("vc * vd =", vc * vd)
print("vc * 2 =", vc * 2)
print("10 * vc =", 10 * vc)
```

## 实验2：函数、条件、循环与递归
### 实验目的
1. 编写`score_level(score)`分数等级函数，实现分级判断，超出0~100抛出`ValueError`异常
2. 分别使用`for`循环、`while`循环计算1~100整数累加，对比结果
3. 编写迭代版、递归版阶乘函数，约定 $0! = 1$，测试 n=0,1,5,10
4. 编写`summary(values)`函数，接收非空数值列表，以元组返回最小、最大、平均值，并使用元组解包
5. 编写`sum_all(*args)`可变参数求和函数，测试无参数、单个参数、多个参数场景


```python
# 任务1：score_level 分数等级函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0到100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

print("=====任务1 分数等级测试=====")
test_scores = [95, 85, 70, 50]
for s in test_scores:
    print(f"{s}分 → {score_level(s)}")

# 任务2：for 和 while 分别求1~100的和
print("\n=====任务2 1~100累加=====")
sum_for = 0
for i in range(1, 101):
    sum_for += i
print(f"for循环结果：{sum_for}")

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print(f"while循环结果：{sum_while}")
print(f"两种结果相等？ {sum_for == sum_while}")

# 任务3：迭代版、递归版阶乘
def fact_iter(n):
    res = 1
    for k in range(2, n+1):
        res *= k
    return res

def fact_recur(n):
    if n == 0:
        return 1
    return n * fact_recur(n - 1)

print("\n=====任务3 阶乘对比=====")
for n in [0,1,5,10]:
    print(f"n={n}，迭代阶乘={fact_iter(n)}，递归阶乘={fact_recur(n)}")

# 任务4 summary函数，返回最小、最大、平均值，元组解包
def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values) / len(values)
    return (min_v, max_v, avg_v)

print("\n=====任务4 summary函数测试=====")
data = [10, 20, 30, 40]
min_res, max_res, avg_res = summary(data) #元组解包
print(f"最小值：{min_res}, 最大值：{max_res}, 平均值：{avg_res:.2f}")

#任务5 sum_all(*args)可变参数
def sum_all(*args):
    return sum(args)

print("\n=====任务5 sum_all可变参数测试=====")
print(f"无参数调用 sum_all() = {sum_all()}")
print(f"单个参数 sum_all(10) = {sum_all(10)}")
print(f"多个参数 sum_all(1,2,3,4) = {sum_all(1,2,3,4)}")
```

## 实验3：列表、字典、元组与基础统计

3. 掌握列表推导式，筛选及格学生、生成成绩平方列表

> 模拟成绩数据：

```python
# 原始模拟数据
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

print("=====任务1 zip构建字典 + items遍历=====")
stu_dict = dict(zip(names, scores))
for name, s in stu_dict.items():
    print(f"{name}: {s}")

print("\n=====任务2 enumerate，序号从1开始=====")
for idx, (name, s) in enumerate(zip(names, scores), start=1):
    print(f"{idx}. {name} 成绩：{s}")

print("\n=====任务3 列表推导式=====")
# 提取及格学生姓名（>=60）
pass_names = [name for name, s in zip(names, scores) if s >= 60]
# 所有成绩平方列表
score_square = [s**2 for s in scores]
print(f"及格学生姓名：{pass_names}")
print(f"成绩平方列表：{score_square}")

print("\n=====任务4 基础统计=====")
count = len(scores)
avg = sum(scores)/count
max_sc = max(scores)
min_sc = min(scores)
pass_cnt = len([s for s in scores if s >=60])
pass_rate = pass_cnt / count

print(f"总人数：{count}")
print(f"平均分：{avg:.2f}")
print(f"最高分：{max_sc}")
print(f"最低分：{min_sc}")
print(f"及格率：{pass_rate:.2%}")

print("\n=====任务5 别名 vs copy() 演示=====")
demo_list = [10,20,30,40]
alias = demo_list
copied = demo_list.copy()

alias[0] = 999
print(f"原demo_list：{demo_list}")
print(f"alias别名：{alias}")
print(f"copied副本：{copied}")
# 结论：alias只是引用，修改alias会改动原列表；copy()生成独立新列表

print("\n=====任务6 元组不可变 + zip不等长序列=====")
# 元组保存学生记录
stu_tuple = ("Alice", 85)
print(f"学生元组记录：{stu_tuple}")
# stu_tuple[1] = 90  # 取消注释会报错：元组元素不可赋值

# 长度不同的两个序列zip
list_a = [1,2,3]
list_b = ["x", "y"]
zip_result = list(zip(list_a, list_b))
print(f"不等长序列zip结果（会截断到短序列长度）：{zip_result}")
```

## 实验4：文件读写、异常处理与模块
### 实验任务
1. 使用`os.getcwd()`查看工作目录，`os.path.join()`构造路径，`os.makedirs(..., exist_ok=True)`创建`exp01_data`文件夹。

4. 逐行读取文件，捕获异常，记录行号+错误信息，继续处理后续行；有效记录存入字典列表。
5. 统计有效/无效记录数、平均分、及格率，写入`exp01_data/report.txt`；演示`w`覆盖模式与`a`追加模式。

> students.txt内容：
name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89

Incomplete


```python
import os
import pickle

# =========任务1：目录操作=========
cwd = os.getcwd()
print(f"当前工作目录: {cwd}")
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"数据目录路径: {data_dir}")

# =========任务2：写入并读取students.txt=========
txt_path = os.path.join(data_dir, "students.txt")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
# 写入
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content)

# 读取
with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
print("原始读取全部行：")
for l in lines:
    print(repr(l.strip()))


# =========任务3：解析函数 parse_student =========
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段，需要2个")
    name, score_str = parts
    name = name.strip()
    if len(name) == 0:
        raise ValueError("姓名不能为空")
    score = int(score_str.strip())
    if not (0 <= score <= 100):
        raise ValueError(f"成绩{score}超出0~100范围")
    return {"name": name, "score": score}


# =========任务4：逐行处理、异常捕获，收集有效记录=========
valid_records = []
invalid_info = []

for line_no, raw_line in enumerate(lines, start=1):
    # 跳过表头第一行
    if line_no == 1:
        continue
    try:
        rec = parse_student(raw_line)
        valid_records.append(rec)
    except Exception as e:
        invalid_info.append({"line": line_no, "error": str(e)})

print("\n====有效记录====")
for r in valid_records:
    print(r)
print("\n====无效记录====")
for bad in invalid_info:
    print(f"行{bad['line']}: {bad['error']}")

# 统计
valid_cnt = len(valid_records)
invalid_cnt = len(invalid_info)
if valid_cnt > 0:
    scores = [r["score"] for r in valid_records]
    avg_score = sum(scores) / valid_cnt
    pass_count = sum(1 for s in scores if s >= 60)
    pass_rate = pass_count / valid_cnt
else:
    avg_score = 0
    pass_rate = 0

print(f"\n有效记录数:{valid_cnt}, 无效记录数:{invalid_cnt}")
print(f"平均分:{avg_score:.2f}, 及格率:{pass_rate:.2%}")


# =========任务5：写入report.txt，演示w模式与a模式=========
report_path = os.path.join(data_dir, "report.txt")
# w模式：覆盖原有文件
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {valid_cnt}\n")
    f.write(f"无效记录数: {invalid_cnt}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")
print("\nw模式写入完成（覆盖旧文件）")

# a模式：追加内容，不会清空原有内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("====追加测试行（a模式）====\n")
print("a模式追加一行完成，不会覆盖原有内容")


# =========任务6 pickle序列化保存与读取=========
pkl_path = os.path.join(data_dir, "students.pkl")
# dump保存
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
# load读取
with open(pkl_path, "rb") as f:
    loaded_records = pickle.load(f)

print("\n====pickle校验，数据是否相等：", loaded_records == valid_records)
print("pickle读取出来的数据：", loaded_records)


# =========任务8：读取不存在文件，捕获FileNotFoundError=========
not_exist_path = os.path.join(data_dir, "no_such_file.txt")
try:
    with open(not_exist_path, "r", encoding="utf-8") as f:
        data = f.read()
except FileNotFoundError as e:
    print(f"\n捕获FileNotFoundError：{e}，文件不存在")
```

## 实验5：类、继承与对象复制
### 操作任务
1. 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()和__str__()方法。
2. 为实验4中的每条有效记录创建一个Student对象，输出对象信息及及格状态。
3. 定义继承自Student的GraduateStudent类，增加research_topic属性，通过super().__init__()初始化父类属性，并重写__str__()。
4. 构造包含嵌套列表的对象，例如`{"name": "Alice", "scores": [85, 90]}`。分别进行直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，比较结果。每次比较都从新建的原始对象开始。


```python
import copy

# =========任务1：定义Student类=========
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        # >=60分为及格
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


# =========任务2：用实验4的有效记录创建Student对象=========
# 实验4得到的有效记录列表
valid_records = [
    {'name': 'Alice', 'score': 85},
    {'name': 'Bob', 'score': 58},
    {'name': 'Carol', 'score': 92},
    {'name': 'David', 'score': 76},
    {'name': 'Eva', 'score': 60},
    {'name': 'Frank', 'score': 89}
]

print("====任务2 Student对象实例====")
student_obj_list = []
for rec in valid_records:
    s = Student(rec["name"], rec["score"])
    student_obj_list.append(s)
    status = "及格" if s.is_passed() else "不及格"
    print(f"{s}，{status}")


# =========任务3：继承，GraduateStudent子类=========
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic})"

print("\n====任务3 研究生子类测试====")
gs = GraduateStudent("LiMing", 88, "大数据分析")
print(gs)
print("是否及格：", gs.is_passed())


# =========任务4：直接赋值、浅拷贝copy.copy、深拷贝copy.deepcopy=========
print("\n====任务4 对象拷贝对比实验====")

# 原始对象
origin = {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值（引用，完全同一个对象）
obj_assign = origin
obj_assign["scores"][0] = 99
print(f"直接赋值后 origin = {origin}")
# 重置原始对象
origin = {"name": "Alice", "scores": [85, 90]}

# 2. 浅拷贝 copy.copy()
obj_shallow = copy.copy(origin)
obj_shallow["scores"][0] = 99
print(f"浅拷贝copy.copy后 origin = {origin}")
origin = {"name": "Alice", "scores": [85, 90]}

#3. 深拷贝 copy.deepcopy()
obj_deep = copy.deepcopy(origin)
obj_deep["scores"][0] = 99
print(f"深拷贝copy.deepcopy后 origin = {origin}")
```

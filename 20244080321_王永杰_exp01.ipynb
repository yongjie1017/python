## 实验2：函数、条件、循环与递归
先写分数判断函数，对非法分数抛出异常。
分别用for和while循环计算1~100累加和，对比两种循环写法。


```python
# 1.分数等级函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0到100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

print(score_level(94))
print(score_level(81))
print(score_level(66))
print(score_level(45))

# 2 for、while实现1-100求和
sum_for = 0
for i in range(1,101):
    sum_for += i
print("for循环求和结果：", sum_for)

sum_while = 0
n = 1
while n <= 100:
    sum_while += n
    n +=1
print("while循环求和结果：", sum_while)
```

继续写阶乘，分别用迭代和递归两种方式实现，0的阶乘规定为1。
还有返回多值的summary函数，以及接收不定长参数的sum_all。


```python
#3 阶乘：迭代写法 和 递归写法
def fact_iter(n):
    res = 1
    for i in range(1, n+1):
        res *= i
    return res

def fact_recur(n):
    if n == 0:
        return 1
    return n * fact_recur(n-1)

for num in [0,1,5,10]:
    print(f"n={num} 迭代:{fact_iter(num)} 递归:{fact_recur(num)}")

#4 summary函数，返回最小、最大、平均值，元组解包接收
def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values)/len(values)
    return min_v, max_v, avg_v

out = summary([12,45,78])
mi, ma, av = out
print("最小，最大，平均：", mi, ma, av)

#5 可变参数 *args
def sum_all(*args):
    return sum(args)

print("无参数：", sum_all())
print("单个参数：", sum_all(9))
print("多个参数：", sum_all(2,4,6,8))
```

## 实验3：列表、字典、元组与基础统计
练习zip把两个列表配对生成字典，enumerate设置从1开始编号。
用列表推导式快速筛选及格同学，做成绩统计。
区分直接赋值和copy()浅拷贝的区别，同时测试元组不可修改、长短列表zip的特点。


```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

#1 zip构造字典
stu_dict = dict(zip(names, scores))
for k,v in stu_dict.items():
    print(k, v)

#2 enumerate序号从1开始
for idx,(name,sc) in enumerate(zip(names,scores), start=1):
    print(idx, name, sc)

#3 列表推导式筛选
pass_name = [name for name,sc in zip(names,scores) if sc>=60]
score_square = [s*s for s in scores]
print("及格同学：", pass_name)
print("成绩平方：", score_square)

#4 统计信息
total = len(scores)
avg_score = sum(scores)/total
max_sc = max(scores)
min_sc = min(scores)
pass_count = len([s for s in scores if s>=60])
pass_rate = pass_count / total
print(f"总人数:{total},平均分:{avg_score:.2f},最高分:{max_sc},最低分:{min_sc},及格率:{pass_rate:.2%}")

#5 拷贝演示，单独用demo列表，不破坏原始数据
demo_list = [100,200,300]
alias = demo_list
copy_list = demo_list.copy()
alias[0] = 999
print("原demo列表：", demo_list)
print("alias直接赋值：", alias)
print("copy副本：", copy_list)

#6 元组、长短不同序列zip测试
stu_tuple = ("Tom", 77)
# stu_tuple[1] = 88  # 取消注释会报错，元组元素不能修改
list_a = [1,2]
list_b = [10,20,30]
print("长短不同list zip结果：", list(zip(list_a, list_b)))
```

## 实验4：文件读写、异常处理与模块
创建exp01_data文件夹。students.txt包含正常记录和几条错误记录。
parse_student写在外部student_utils.py，通过import导入。
按行解析，捕获错误记录，统计有效、无效记录。演示w和a模式，pickle保存读取数据。


```python
import os
import pickle
from student_utils import parse_student

# 1. 创建exp01_data文件夹
work_dir = os.getcwd()
print("当前工作目录：", work_dir)
data_folder = os.path.join(work_dir, "exp01_data")
os.makedirs(data_folder, exist_ok=True)
txt_file = os.path.join(data_folder, "students.txt")

# 2. 写入带错误数据的students.txt
text_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(txt_file, "w", encoding="utf-8") as f:
    f.write(text_content)

# 3. 逐行读取解析，捕获异常
valid_data = []
error_num = 0
valid_num = 0

with open(txt_file,"r",encoding="utf-8") as f:
    all_lines = f.readlines()

for line_index, one_line in enumerate(all_lines, start=1):
    if line_index == 1:
        continue  # 跳过表头
    try:
        rec = parse_student(one_line)
        valid_data.append(rec)
        valid_num += 1
    except ValueError as e:
        error_num += 1
        print(f"第{line_index}行出错：{e}")

# 统计平均分、及格率
if valid_num>0:
    average = sum(x["score"] for x in valid_data)/valid_num
    pass_stu = sum(1 for x in valid_data if x["score"]>=60)
    pass_ratio = pass_stu / valid_num
else:
    average =0
    pass_ratio=0

# 4. w模式写入report.txt（覆盖原有内容）
report_path = os.path.join(data_folder, "report.txt")
report_text = f"""有效记录数：{valid_num}
无效记录数：{error_num}
平均分：{average:.2f}
及格率：{pass_ratio:.2%}
"""
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report_text)
print("w模式写入完成，w会清空文件原有全部内容")

# a模式追加内容（不覆盖）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("-----a模式测试追加的一行-----\n")
print("a模式追加完成，a在末尾添加，不会删除旧内容")

# 5. pickle序列化保存与读取
pkl_path = os.path.join(data_folder, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_data, f)

with open(pkl_path, "rb") as f:
    load_back = pickle.load(f)
print("pickle读取后和原数据相等：", load_back == valid_data)

# 测试外部导入的parse_student
print("\n测试外部模块函数：")
print("有效样例：", parse_student("Lily,82"))
try:
    parse_student("BadMan,200")
except ValueError as e:
    print("无效样例报错：", e)

# 捕获文件不存在异常
try:
    with open("no_such_file.txt","r",encoding="utf-8") as ff:
        pass
except FileNotFoundError:
    print("\n捕获异常：目标文件不存在")
```

## 实验5：面向对象、类继承与深浅拷贝
定义父类Student，包含__init__、is_passed、__str__方法；
子类GraduateStudent继承Student，新增导师属性，重写__str__；
对比直接赋值、浅拷贝copy.copy、深拷贝copy.deepcopy的区别。


```python
import copy

class Student:
    def __init__(self, name, score, info_list):
        self.name = name
        self.score = score
        self.info_list = info_list

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, info={self.info_list})"


class GraduateStudent(Student):
    def __init__(self, name, score, info_list, supervisor):
        super().__init__(name, score, info_list)
        self.supervisor = supervisor

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, supervisor={self.supervisor}, info={self.info_list})"


# 创建对象
s1 = Student("Alice", 85, ["sport", "music"])
print("原始对象 s1：", s1)
print("Alice 是否及格：", s1.is_passed())

g1 = GraduateStudent("Bob", 55, ["coding"], "Prof.Wang")
print("研究生对象 g1：", g1)
print("Bob 是否及格：", g1.is_passed())

# 三种拷贝对比
s_assign = s1          # 直接赋值：同一个对象
s_shallow = copy.copy(s1)  # 浅拷贝
s_deep = copy.deepcopy(s1)  # 深拷贝

# 修改内部列表元素，观察差异
s1.info_list.append("reading")
print("\n=== 修改s1.info_list之后 ===")
print("s1:         ", s1)
print("直接赋值s_assign：", s_assign)
print("浅拷贝s_shallow：  ", s_shallow)
print("深拷贝s_deep：     ", s_deep)
```

## 热身：定义简单函数
1. 定义一个名为 `say_hi` 的函数，它不接受任何参数，并在被调用时打印字符串 "Hello, world!"。
2. 定义一个名为 `goat_pad` 的函数，它接受一个字符串作为其唯一参数，并打印该字符串，前后都加上字符串 "goat"。
3. 定义一个名为 `print_n` 的函数，它接受两个参数，一个字符串 `s` 和一个整数 `n`，并打印字符串 `n` 次，每次在单独的一行上。


```python
# ## 热身：定义简单函数
# 1. say_hi：无参函数，打印 Hello, world!
# 2. goat_pad：接收1个字符串参数，在字符串前后拼接goat后打印
# 3. print_n：接收字符串s和整数n，单独每行打印s，一共打印n次

def say_hi():
    # 不接收参数，调用时打印"Hello, world!"
    print("Hello, world!")

def goat_pad(s):
    # 参数 s: 输入的字符串
    # 在s的前面和后面都加上"goat"，然后打印
    print("goat" + s + "goat")

def print_n(s, n):
    # 参数 s: 需要打印的字符串
    # 参数 n: 打印次数
    # 循环n次，每次单独一行打印s
    for _ in range(n):
        print(s)


# ===== 测试代码 =====
print("=== 测试 say_hi() ===")
say_hi()

print("\n=== 测试 goat_pad() ===")
goat_pad("test")

print("\n=== 测试 print_n() ===")
print_n("hi", 3)
```

## 欧几里得算法
欧几里得算法是一种寻找两个数的最大公约数（GCD）的方法。两个数m和n的最大公约数是同时能整除m和n的最大数。

1. 实现一个名为`gcd`的函数，它接受两个整数作为其参数并返回它们的GCD。假设两个输入都是整数，不需要做错误检查。
2. 使用函数计算下面几组数字的GCD：
    1. (2018, 2019)
    2. (1200, 300)
    3. (5040, 60)


```python
# 欧几里得算法：求两个整数的最大公约数GCD
def gcd(m, n):
    # 循环：当n不等于0，持续迭代
    while n != 0:
        # m更新为n，n更新为m对n取模的余数
        m, n = n, m % n
    # 余数为0时，m就是最大公约数，返回
    return m


# 题目要求的三组测试
print("=== 题目指定测试用例 ===")
print(f"gcd(2018, 2019) = {gcd(2018, 2019)}")
print(f"gcd(1200, 300) = {gcd(1200, 300)}")
print(f"gcd(5040, 60) = {gcd(5040, 60)}")

# 额外测试负数情况，回答第3小问
print("\n=== 测试负数参数 ===")
print(f"gcd(-12, 8) = {gcd(-12, 8)}")
print(f"gcd(-12, -8) = {gcd(-12, -8)}")
```

## 近似欧拉数e
自然对数的底数e通常被定义为无限和
$$
e=\sum_{k=0}^{\infty}\frac{1}{k!}=1+\frac11+\frac12+\frac16+\frac{1}{24}+\dots
$$
其中$k!$表示k的阶乘，约定$0!=1$。

1. 定义`euler_limit(n)`：接收整数n，使用极限公式 $\big(1+\frac1n\big)^n$ 近似e，返回浮点数。
2. 定义`euler_infinite_sum(n)`：接收非负整数n，计算公式(1)前n项之和近似e。
    - $n=0$ 返回0
    - $n=1$ 返回1（k=0这一项）
    - $n=4$ 计算前4项：$1+\frac11+\frac12+\frac16 \approx 2.6667$
3. 定义`euler_approx(ε)`：接收浮点数ε，用while循环累加级数，直到近似值与真实e的误差小于ε，返回近似e。可用`math.exp(1)`获取e真实值。
4. 定义`print_euler_sum_table(n)`：接收正整数n，打印$k=1$到$k=n$对应的`euler_infinite_sum(k)`，每行输出一个值。


```python
import math

# 1. 极限形式近似 e: (1 + 1/n)^n
def euler_limit(n):
    return (1 + 1 / n) ** n


# 2. 无穷级数前n项和近似e
def euler_infinite_sum(n):
    # n=0，返回0
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # 保存k!，初始0! =1
    for k in range(n):
        total += 1 / fact
        fact = fact * (k + 1)
    return total


# 3. 累加级数直到误差小于epsilon
def euler_approx(epsilon):
    e_true = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(e_true - total) > epsilon:
        total += 1 / fact
        k += 1
        fact = fact * k
    return total


# 4. 打印k从1~n的euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        val = euler_infinite_sum(k)
        print(val)


# ========== 测试代码 ==========
print("=== 测试 euler_limit ===")
print(f"euler_limit(10000) = {euler_limit(10000)}")

print("\n=== 测试 euler_infinite_sum ===")
print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}")
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")

print("\n=== 测试 euler_approx，误差小于1e-5 ===")
print(f"euler_approx(1e-5) = {euler_approx(1e-5)}")
print(f"真实e = {math.exp(1)}")

print("\n=== 测试 print_euler_sum_table(6) ===")
print_euler_sum_table(6)
```

## 测试整数属性
一个正整数n是2的幂，如果 $n=2^p$，其中整数 $p\ge0$。

1. 编写`is_power_of_2(n)`：接收正整数n，返回布尔值判断n是否为2的幂。
    - 只能用除法、取模`%`，**不能使用math.sqrt**。
2. 推广为`is_power(b, n)`：接收两个正整数b,n；如果n是b的幂，返回True，否则False。


```python
# 1. 判断n是否是2的幂，仅使用除法和取模
def is_power_of_2(n):
    # 不断除以2，直到不能被2整除
    while n % 2 == 0:
        n = n // 2
    # 最后结果等于1，说明原数是2的幂
    return n == 1


# 2. 判断n是否是b的幂
def is_power(b, n):
    # 当n能被b整除，持续除以b
    while n % b == 0:
        n = n // b
    # 最后剩下1，则n是b的幂
    return n == 1


# 测试代码
print("=== 测试 is_power_of_2 ===")
print(f"is_power_of_2(1)  = {is_power_of_2(1)}")   # 2^0 =1 → True
print(f"is_power_of_2(2)  = {is_power_of_2(2)}")   # True
print(f"is_power_of_2(8)  = {is_power_of_2(8)}")   # True
print(f"is_power_of_2(10) = {is_power_of_2(10)}")  # False

print("\n=== 测试 is_power(b,n) ===")
print(f"is_power(3, 81)  = {is_power(3, 81)}")   # 3^4=81 → True
print(f"is_power(5, 125) = {is_power(5, 125)}")  # 5^3=125 → True
print(f"is_power(2, 12)  = {is_power(2, 12)}")   # False
print(f"is_power(7, 1)   = {is_power(7, 1)}")    # b^0 =1 → True
```

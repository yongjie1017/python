# 热身：定义简单函数
实现3个简单函数：say_hi无参打印hello world；goat_pad前后拼接goat；print_n重复打印字符串n次。


```python
#1 say_hi 不带参数
def say_hi():
    print("Hello, world!")

#2 goat_pad，前后加goat
def goat_pad(s):
    print("goat" + s + "goat")

#3 print_n，字符串s打印n次，每行一个
def print_n(s, n):
    for _ in range(n):
        print(s)

# 测试调用
print("-----测试say_hi-----")
say_hi()

print("\n-----测试goat_pad-----")
goat_pad("bird")
goat_pad("_")

print("\n-----测试print_n-----")
print_n("testline", 3)
```

# 欧几里得算法（GCD最大公约数）
辗转相除法：gcd(a,b)=gcd(b, a mod b)，直到余数为0。
测试三组数字：(2018,2019)、(1200,300)、(5040,60)
同时观察输入负数会出现什么现象。


```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# 题目指定测试样例
print("gcd(2018, 2019):", gcd(2018, 2019))
print("gcd(1200, 300):", gcd(1200, 300))
print("gcd(5040, 60):", gcd(5040, 60))

# 测试负数输入
print("\n测试负数参数：")
print("gcd(-1200,300):", gcd(-1200,300))
print("gcd(-24, -18):", gcd(-24, -18))
```

# 近似计算欧拉数 e
1. euler_limit：极限公式 (1+1/n)^n
2. euler_infinite_sum：级数求和前n项 $ \sum_{k=0}^{n-1} \frac{1}{k!} $，约定0!=1
3. euler_approx：循环累加直到与真实e误差小于ϵ
4. print_euler_sum_table：打印k从1~n的求和结果
使用math.exp(1)拿到e真实值做对比。


```python
import math

#1 极限形式近似e
def euler_limit(n):
    return (1 + 1 / n) ** n

#2 级数前n项和，n=1返回1，n=0返回0
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # 0! = 1
    for k in range(n):
        if k == 0:
            term = 1.0 / fact
        else:
            fact = fact * k
            term = 1.0 / fact
        total += term
    return total

#3 循环迭代，误差小于epsilon停止
def euler_approx(epsilon):
    real_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while True:
        if k == 0:
            term = 1.0 / fact
        else:
            fact = fact * k
            term = 1.0 / fact
        total += term
        if abs(total - real_e) < epsilon:
            break
        k += 1
    return total

#4 打印表格 k=1到n
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

# 测试
print("真实e =", math.exp(1))
print("euler_limit(10000):", euler_limit(10000))
print("euler_infinite_sum(4):", euler_infinite_sum(4))
print("euler_approx(0.001):", euler_approx(0.001))
print("\n打印表格：")
print_euler_sum_table(6)
```

# 判断整数是否为幂
is_power_of_2：判断n是不是2的幂，只用%和//，禁止sqrt。
is_power(b,n)：泛化版本，判断n是否是b的幂。
思路：循环把n不断除以b，如果中途不能整除就False，最后等于1就是True。


```python
def is_power_of_2(n):
    # n是正整数，只用除法和取模
    while n % 2 == 0:
        n = n // 2
    return n == 1

def is_power(b, n):
    while n % b == 0:
        n = n // b
    return n == 1

# 测试样例
print("is_power_of_2(1):", is_power_of_2(1))
print("is_power_of_2(8):", is_power_of_2(8))
print("is_power_of_2(10):", is_power_of_2(10))

print("\nis_power测试：")
print("is_power(3, 9):", is_power(3,9))
print("is_power(5,125):", is_power(5,125))
print("is_power(2,12):", is_power(2,12))
```
